```python
inventory = {"The Great Gatsby": {"price": 10.99, "quantity": 5}, "Rich dad poor dad": {"price": 8.99, "quantity": 8}, "Harry Potter": {"price": 12.50, "quantity": 2}}
```

```python
print(inventory)
```

```python
inventory.update({"Atomic Habits": {"price": 15.99, "quantity": 7}})
```

```python
print(inventory)
```

```python
print(list(inventory.keys()))
```

```python
for book_name in list(inventory.keys()):
    if inventory[book_name]["quantity"] == 0:
        inventory.pop(book_name)
```

```python
print(inventory)
```

```python
letter = input("Enter a single letter: ")
```

```python
if len(letter) == 1 and letter.isalpha():
    if letter.lower() in "aeiou":
        print(letter, "is a Vowel")
    else:
        print(letter, "is a Consonant")
else:
    print("Please enter valid alphabet")
```
